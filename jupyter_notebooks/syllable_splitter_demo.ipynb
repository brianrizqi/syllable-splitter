# Indonesian Syllable Splitter Demo

Notebook ini mendemonstrasikan penggunaan Indonesian Syllable Splitter untuk memisahkan kata bahasa Indonesia menjadi suku kata.

## Setup

Pertama, kita perlu mengimport modul yang diperlukan. Pastikan Anda menjalankan notebook ini dari folder `jupyter_notebooks`.

```python
import sys
import os

# Tambahkan parent directory ke Python path
sys.path.append(os.path.dirname(os.getcwd()))

# Import syllable splitter modules
from HybridSyllableSplitter import HybridSyllableSplitter
from PUEBISyllableSplitter import PUEBISyllableSplitter
from KBBISyllableSplitter import KBBISyllableSplitter
```

## Inisialisasi Syllable Splitter

Kita akan menggunakan `HybridSyllableSplitter` yang merupakan kombinasi dari berbagai metode.

```python
# Inisialisasi splitter
splitter = HybridSyllableSplitter()
print("Syllable Splitter berhasil diinisialisasi!")
```

## Contoh Penggunaan

### 1. Memisahkan Kata Tunggal

```python
# Contoh kata sederhana
kata = "makan"
hasil = splitter.split(kata)
print(f"Kata: {kata}")
print(f"Hasil: {hasil}")
print(f"Suku kata: {'-'.join(hasil)}")
```

### 2. Memisahkan Beberapa Kata Sekaligus

```python
# Daftar kata untuk diuji
kata_kata = [
    "membaca",
    "menulis",
    "belajar",
    "sekolah",
    "perpustakaan",
    "mengemban",
    "memakai",
    "mengetik"
]

print("Hasil pemisahan suku kata:\n")
for kata in kata_kata:
    hasil = splitter.split(kata)
    print(f"{kata:15} → {'-'.join(hasil)}")
```

### 3. Analisis Detail dengan Informasi Morfologi

```python
# Kata dengan imbuhan kompleks
kata_kompleks = "mempertanggungjawabkan"
hasil = splitter.split(kata_kompleks)

print(f"Kata: {kata_kompleks}")
print(f"Jumlah suku kata: {len(hasil)}")
print(f"Suku kata: {'-'.join(hasil)}")
print(f"\nDetail per suku kata:")
for i, suku in enumerate(hasil, 1):
    print(f"  {i}. {suku}")
```

### 4. Membandingkan Metode Berbeda

```python
# Inisialisasi splitter lain
puebi_splitter = PUEBISyllableSplitter()
kbbi_splitter = KBBISyllableSplitter()

kata_test = "pembelajaran"

print(f"Kata: {kata_test}\n")
print(f"Hybrid Splitter: {'-'.join(splitter.split(kata_test))}")
print(f"PUEBI Splitter:  {'-'.join(puebi_splitter.split(kata_test))}")
print(f"KBBI Splitter:   {'-'.join(kbbi_splitter.split(kata_test))}")
```

### 5. Uji Kasus Peluluhan (Nasal Assimilation)

```python
# Kata-kata dengan peluluhan
kata_peluluhan = [
    "memisah",    # me-m-pi-sah (dari pisah)
    "memakai",    # me-m-pa-kai (dari pakai)
    "mengetik",   # me-ng-ke-tik (dari ketik)
    "mengemban",  # me-ng-em-ban (dari emban)
    "menyapu",    # me-ny-sa-pu (dari sapu)
]

print("Uji Kasus Peluluhan:\n")
for kata in kata_peluluhan:
    hasil = splitter.split(kata)
    print(f"{kata:12} → {'-'.join(hasil)}")
```

### 6. Input Interaktif

```python
# Masukkan kata Anda sendiri
kata_input = input("Masukkan kata yang ingin dipisahkan: ")
hasil = splitter.split(kata_input)

print(f"\nHasil pemisahan suku kata:")
print(f"Kata: {kata_input}")
print(f"Suku kata: {'-'.join(hasil)}")
print(f"Jumlah suku kata: {len(hasil)}")
```

## Analisis Batch dari File

```python
# Contoh membaca dari list dan menyimpan hasil
import pandas as pd

# Buat dataframe dengan contoh kata
data = {
    'kata': ['membaca', 'menulis', 'belajar', 'mengajar', 'perpustakaan'],
}

df = pd.DataFrame(data)

# Tambahkan kolom hasil pemisahan
df['suku_kata'] = df['kata'].apply(lambda x: '-'.join(splitter.split(x)))
df['jumlah_suku'] = df['kata'].apply(lambda x: len(splitter.split(x)))

print("Hasil Analisis Batch:\n")
print(df.to_string(index=False))
```

## Export Hasil ke CSV

```python
# Simpan hasil ke CSV
output_file = 'hasil_pemisahan_suku_kata.csv'
df.to_csv(output_file, index=False)
print(f"Hasil berhasil disimpan ke: {output_file}")
```
