# Indonesian Syllable Splitter Demo

Notebook ini mendemonstrasikan penggunaan Indonesian Syllable Splitter untuk memisahkan kata bahasa Indonesia menjadi suku kata.

**Dua metode yang tersedia:**
1. **Local Splitter** (Hybrid, PUEBI) - Offline, cepat
2. **KBBI Scraper** - Online, data resmi dari KBBI

## Setup

Pertama, kita perlu mengimport modul yang diperlukan. Pastikan Anda menjalankan notebook ini dari folder `jupyter_notebooks`.

```python
import sys
import os
import time

# Tambahkan parent directory ke Python path
sys.path.append(os.path.dirname(os.getcwd()))

# Import syllable splitter modules
from HybridSyllableSplitter import HybridSyllableSplitter
from PUEBISyllableSplitter import PUEBISyllableSplitter
from KBBIScraper import KBBIScraper  # Untuk scraping langsung dari KBBI online
```

## Inisialisasi Syllable Splitters

Kita akan menggunakan berbagai splitter untuk perbandingan.

```python
# Inisialisasi splitter lokal (offline)
hybrid_splitter = HybridSyllableSplitter()
puebi_splitter = PUEBISyllableSplitter()

# Inisialisasi KBBI scraper (online - butuh internet)
kbbi_scraper = KBBIScraper()

print("✓ Semua Syllable Splitter berhasil diinisialisasi!")
print(f"  - Hybrid Splitter (offline)")
print(f"  - PUEBI Splitter (offline)")
print(f"  - KBBI Scraper (online) - URL: {kbbi_scraper.base_url}")
```

---
# BAGIAN 1: Local Splitter (Offline)

## 1. Memisahkan Kata Tunggal

```python
# Contoh kata sederhana
kata = "makan"
hasil = hybrid_splitter.split_syllables(kata)
print(f"Kata: {kata}")
print(f"Hasil: {hasil}")
print(f"Suku kata: {'-'.join(hasil)}")
```

## 2. Memisahkan Beberapa Kata Sekaligus

```python
# Daftar kata untuk diuji
kata_kata = [
    "membaca",
    "menulis",
    "belajar",
    "sekolah",
    "perpustakaan",
    "mengemban",
    "memakai",
    "mengetik"
]

print("Hasil pemisahan suku kata (Hybrid Splitter):\n")
for kata in kata_kata:
    hasil = hybrid_splitter.split_syllables(kata)
    print(f"{kata:15} → {'-'.join(hasil)}")
```

## 3. Analisis Detail dengan Informasi Morfologi

```python
# Kata dengan imbuhan kompleks
kata_kompleks = "mempertanggungjawabkan"
hasil = hybrid_splitter.split_syllables(kata_kompleks)

print(f"Kata: {kata_kompleks}")
print(f"Jumlah suku kata: {len(hasil)}")
print(f"Suku kata: {'-'.join(hasil)}")
print(f"\nDetail per suku kata:")
for i, suku in enumerate(hasil, 1):
    print(f"  {i}. {suku}")
```

## 4. Membandingkan Metode Local Splitter

```python
kata_test = "pembelajaran"

print(f"Kata: {kata_test}\n")
print(f"Hybrid Splitter: {'-'.join(hybrid_splitter.split_syllables(kata_test))}")
print(f"PUEBI Splitter:  {'-'.join(puebi_splitter.split_syllables(kata_test))}")
```

## 5. Uji Kasus Peluluhan (Nasal Assimilation)

```python
# Kata-kata dengan peluluhan
kata_peluluhan = [
    "memisah",    # me-m-pi-sah (dari pisah)
    "memakai",    # me-m-pa-kai (dari pakai)
    "mengetik",   # me-ng-ke-tik (dari ketik)
    "mengemban",  # me-ng-em-ban (dari emban)
    "menyapu",    # me-ny-sa-pu (dari sapu)
]

print("Uji Kasus Peluluhan (Hybrid Splitter):\n")
for kata in kata_peluluhan:
    hasil = hybrid_splitter.split_syllables(kata)
    print(f"{kata:12} → {'-'.join(hasil)}")
```

---
# BAGIAN 2: KBBI Scraper (Online)

**Catatan:** Bagian ini membutuhkan koneksi internet karena mengambil data langsung dari website KBBI.

## 6. Mengambil Data dari KBBI Online

```python
# Contoh mengambil satu kata dari KBBI
kata = "pembelajaran"
print(f"Mencari kata: {kata}")
print(f"URL: {kbbi_scraper.base_url}/entri/{kata}\n")

hasil = kbbi_scraper.get_syllables(kata)

if hasil:
    print(f"✓ Berhasil!")
    print(f"Kata: {kata}")
    print(f"Suku kata dari KBBI: {'-'.join(hasil)}")
    print(f"Jumlah suku: {len(hasil)}")
else:
    print(f"✗ Kata '{kata}' tidak ditemukan di KBBI")
```

## 7. Batch Processing dari KBBI

**Perhatian:** Scraping membutuhkan waktu. Gunakan delay untuk tidak membebani server KBBI.

```python
# Daftar kata untuk diambil dari KBBI
kata_kbbi = [
    "membaca",
    "menulis",
    "belajar",
    "sekolah",
    "perpustakaan"
]

print("Mengambil data dari KBBI online...\n")
print("="*60)

for kata in kata_kbbi:
    hasil = kbbi_scraper.get_syllables(kata)
    
    if hasil:
        print(f"✓ {kata:15} → {'-'.join(hasil)}")
    else:
        print(f"✗ {kata:15} → Tidak ditemukan")
    
    # Delay untuk tidak membebani server
    time.sleep(0.5)

print("="*60)
```

## 8. Perbandingan: KBBI Scraper vs Local Splitter

```python
# Kata untuk dibandingkan
kata_compare = [
    "pembelajaran",
    "mengetik",
    "perpustakaan",
    "memakai"
]

print("Perbandingan KBBI Scraper (Online) vs Hybrid Splitter (Offline):\n")
print("="*80)

for kata in kata_compare:
    # Dari KBBI Online
    kbbi_result = kbbi_scraper.get_syllables(kata)
    kbbi_str = '-'.join(kbbi_result) if kbbi_result else "Tidak ditemukan"
    
    # Dari Hybrid Splitter (Offline)
    hybrid_result = hybrid_splitter.split_syllables(kata)
    hybrid_str = '-'.join(hybrid_result)
    
    # Cek apakah sama
    sama = "✓ SAMA" if kbbi_result == hybrid_result else "✗ BEDA"
    
    print(f"Kata: {kata}")
    print(f"  KBBI Online:      {kbbi_str}")
    print(f"  Hybrid (Offline): {hybrid_str}")
    print(f"  Status: {sama}")
    print("-"*80)
    
    time.sleep(0.5)
```

## 9. Input Interaktif

```python
# Masukkan kata Anda sendiri
kata_input = input("Masukkan kata yang ingin dipisahkan: ")

print(f"\n{'='*70}")
print(f"Hasil pemisahan untuk kata: {kata_input}")
print(f"{'='*70}\n")

# Hybrid Splitter (Offline)
hybrid_result = hybrid_splitter.split_syllables(kata_input)
print(f"Hybrid Splitter (Offline): {'-'.join(hybrid_result)}")

# KBBI Scraper (Online)
print(f"\nMengambil dari KBBI online...")
kbbi_result = kbbi_scraper.get_syllables(kata_input)
if kbbi_result:
    print(f"KBBI Scraper (Online):     {'-'.join(kbbi_result)}")
    print(f"\nURL KBBI: {kbbi_scraper.base_url}/entri/{kata_input}")
else:
    print(f"KBBI Scraper (Online):     Tidak ditemukan di KBBI")

print(f"\n{'='*70}")
```

## 10. Analisis Batch dengan Pandas

```python
import pandas as pd

# Buat dataframe dengan contoh kata
data = {
    'kata': ['membaca', 'menulis', 'belajar', 'mengajar', 'perpustakaan'],
}

df = pd.DataFrame(data)

# Tambahkan kolom hasil pemisahan dari Hybrid Splitter (cepat, offline)
df['hybrid_splitter'] = df['kata'].apply(lambda x: '-'.join(hybrid_splitter.split_syllables(x)))
df['jumlah_suku'] = df['kata'].apply(lambda x: len(hybrid_splitter.split_syllables(x)))

print("Hasil Analisis Batch (Hybrid Splitter):\n")
print(df.to_string(index=False))
```

# Kesimpulan

## Perbandingan Metode

### Local Splitter (Hybrid, PUEBI)
**Kelebihan:**
- ✓ Sangat cepat (offline)
- ✓ Tidak butuh internet
- ✓ Cocok untuk batch processing besar
- ✓ Selalu tersedia

**Kekurangan:**
- ✗ Mungkin tidak 100% akurat untuk semua kasus
- ✗ Perlu update manual jika ada perubahan aturan

### KBBI Scraper (Online)
**Kelebihan:**
- ✓ Data resmi dari KBBI
- ✓ Selalu up-to-date
- ✓ Akurat sesuai standar KBBI

**Kekurangan:**
- ✗ Butuh koneksi internet
- ✗ Lebih lambat (network request)
- ✗ Tidak semua kata ada di KBBI
- ✗ Bisa gagal jika server down

## Rekomendasi Penggunaan

1. **Untuk processing cepat & batch besar:** Gunakan **Hybrid Splitter**
2. **Untuk validasi & referensi resmi:** Gunakan **KBBI Scraper**
3. **Best practice:** Gunakan Hybrid Splitter dengan KBBI Scraper sebagai validasi untuk kata-kata penting
